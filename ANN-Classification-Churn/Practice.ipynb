```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Churn_Modelling.csv")
data.head(5)
input_data = data.drop(["RowNumber", "CustomerId", "Surname"], axis=1)
input_data
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
##Lable Encode the Gender data
label_encoder=LabelEncoder()
input_data['Gender']=label_encoder.fit_transform(input_data["Gender"])
print(input_data)



```

```python
## One Hot Encoding for geodata
one_hot_encoder = OneHotEncoder()
geo_encoded = one_hot_encoder.fit_transform(input_data[["Geography"]]).toarray()
column_names = one_hot_encoder.get_feature_names_out(["Geography"])
```

```python
input_data = input_data.drop(["Geography"], axis=1)
input_data
```

```python
# convert encoded data to dataframe
encoded_df = pd.DataFrame(geo_encoded, columns=column_names, index=input_data.index)

##Combine the data
input_data_df = pd.concat([input_data, encoded_df], axis=1)
input_data_df
```

```python
##Split data into dependent and independent parameters

X = input_data_df.drop("EstimatedSalary",axis=1)
y = input_data_df["EstimatedSalary"]

print(X)
print(y)
```

```python
# Split data into training and testing set
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```
