```python
import pandas as pd
import numpy as np
```

```python
from tensorflow.keras.preprocessing.text import one_hot
sent = [
    "Machine learning is transforming many industries today.",
    "Artificial intelligence helps computers learn from data.",
    "Deep learning uses neural networks with many layers.",
    "Natural language processing enables computers to understand text.",
    "Word embeddings capture semantic relationships between words.",
    "Data scientists analyze large datasets to discover insights.",
    "Python is a popular programming language for machine learning.",
    "TensorFlow and PyTorch are widely used deep learning frameworks.",
    "Good data quality improves the performance of machine learning models.",
    "Technology continues to evolve and create new opportunities."
]
```

```python
sent
```

```python

vocab_size=10000
one_hot=[one_hot(words, vocab_size)for words in sent]
one_hot
```

```python
##Word Enbeding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
##Padding of the wrords 

sent_length= 10
embeded_doc=pad_sequences(one_hot, padding='pre', maxlen=10)
embeded_doc

for doc in embeded_doc:
    print(f"Only 1 doc is printed: {doc}")
    break
```

Create Model and add embedding  layer

```python
from tensorflow.keras.layers import Input
from tensorflow.keras.models import Sequential
dim=10
model=Sequential([Input(shape=(sent_length,)),])
model.add(Embedding(vocab_size,dim ,input_length=sent_length))
model.summary()
model.predict(embeded_doc)

model.compile(optimizer='adam', loss='mse')
```
